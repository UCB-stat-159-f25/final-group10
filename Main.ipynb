## Introduction

Redlining refers to the practice of grading neighborhoods by perceived lending risk in the mid-twentieth century, most notably through maps produced by the Home Owners’ Loan Corporation (HOLC) in the 1930s. These maps classified neighborhoods from “A” (Best) to “D” (Hazardous), with areas graded C and D disproportionately associated with communities of color and immigrant populations. Though redlining polices were outlawed decades ago, research show that the contemporary housing market are still in the shadow of the redlining's spatial and economic consequences. 

This project investigate whether pattern associated with historical redlining continue to be reflected in present-day housing and demographic outcomes. Using national HOLC data and county-level American Community Survey (ACS) estimates, the analysis situates redlining in historical context and evaluates its persistence in the Bay Area, focusing on Alameda (Oakland), Santa Clara (San Jose), and San Francisco counties.

### Goal of Analysis

The main goal of this analysis is to assess whether historically redlined area exhibit presistent disparities in houseing and demographic composition. The project address three main questions 
1. How do housing value and demographic differ across HOLC grades at the national level? 
2. Do similar patterns appear within Bay Area counties?
3. Is there evidence that disparities between redlined and non-redlined areas have narrowed over time?

### Data and Methodological Overview

- **Historical Redlining (HOLC):** Digitized Home Owners’ Loan Corporation (HOLC)
  neighborhood grades were obtained from the
  [Mapping Inequality Project](https://dsl.richmond.edu/panorama/redlining/#loc=5/39.1/-94.58),
  which provides spatial data on historical redlining practices in U.S. cities.

- **National Redlining Time Series:** National-level redlining-related housing
  indicators were accessed through the Federal Reserve Bank of St. Louis’
  [FRED database](https://fred.stlouisfed.org/tags/series?t=redlining),
  which aggregates historical economic and housing time series.

- **Contemporary Housing and Demographics:** Tract-level housing and demographic
  data were drawn from the U.S. Census Bureau’s
  [American Community Survey (ACS) 5-Year Estimates](https://data.census.gov/),
  covering the period 2009–2022. Included data table: Race(B03002), Income(B19013), Median Home Value(B25077), Homeownerhip(B25003)



Methodologically, the project combines spatial analysis, visualization and regression-based comparison. National time-series analyses provide historical context, while Bay area case studies link HOLC grades to current day outcomes. Difference-in-differece comparsion and regression are used to examine whether the inequality presisted over time. the result are intrepreted as associational evidence rather than casual effects.

## Persistence Over Time — Difference-in-Differences Evidence

To assess whether disparities associated with redlining have narrowed over time, we employ difference-in-differences–style comparisons between historically favored (Grades A/B) and disfavored (Grades C/D) areas. Figure 10 shows parallel trends prior to key policy changes, supporting the use of longitudinal comparison.

![p](figures/nation_wide/home_values_parallel_check.png)

**Figure 9**: parallel trends prior to key policy change
| *Source*: @fred_redlining

```python
import pandas as pd
from src.did_utils import fit_did_ols

home_long = pd.read_csv("data/home_long.csv") 

did_model_home = fit_did_ols(
    home_long,
    outcome_col="home_value",
    treat_col="treatment",
    post_col="post_1968",
    interaction_col="treat_post",
)

print(did_model_home.summary())
```

The difference-in-differences analysis finds no statistically significant effect of the 1968 Fair Housing Act on reducing the home value gap between historically redlined (Grade D) and non-redlined (Grade A) neighborhoods. The interaction term (treat_post) is small and statistically insignificant, indicating that post-1968 home value trends in redlined areas did not converge toward those of Grade A neighborhoods beyond the pre-existing gap. In practice, this suggests that redlined neighborhoods remained substantially lower in median home value even decades after the policy’s enactment. These results imply that although the Fair Housing Act formally prohibited discriminatory lending practices, legislative change alone was insufficient to reverse the long-term economic effects of redlining and the entrenched patterns of disinvestment it produced.

## Conclusion 
Together, the national and Bay Area analyses suggest that historical redlining is closely associated with long-lasting housing and demographic disparities. While overall housing values have increased, relative gaps between historically redlined and non-redlined areas remain. These findings highlight how historical policy decisions can shape urban inequality across generations.

## Limitation and Assumption

This analysis relies on historical HOLC redlining maps and publicly available ACS and housing value data, over which we have no control in terms of original data collection or measurement. We assume that these datasets are internally consistent across time and geography, and that any measurement error is not systematically correlated with treatment status. Given the historical nature of redlining, we treat HOLC grades as fixed indicators of past institutional exposure rather than precise measures of individual-level discrimination.

Our difference-in-differences framework assumes that, absent the policy intervention, treated (historically redlined) and control (non-redlined) areas would have followed parallel trends in housing values. This assumption is partially assessed through pre-policy trend comparisons. While this analysis does not establish causal effects definitively, it allows us to explore whether large-scale anti-discrimination legislation coincided with differential changes in housing outcomes across historically distinct neighborhoods. Given the exploratory scope of the project, we focus on broad, interpretable patterns rather than optimizing model specification or making strong causal claims.
